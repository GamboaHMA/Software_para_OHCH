Resumiendo, ya podemos identificar los tipos de palabras del idioma espaniol dado un texto, ahora sobre lo que nos devolvio el OCR tenemos que 
hacer una limpieza, la cual incluye quitar espacios extra, ya sea porque dividen a palabras o espacios entre linea y linea

```python
import re # expresiones regulares 

def limpiar_espacios_basicos(text:str) -> str:
    # 1. Unir cortes de linea con guion al final "pala-\nbra" -> "palabra"
    text = re.sub(r'(\w)-\s*\n\s*(\w)', r'\1\2', text)

    text = text.replace('-', ' ')

    # 2. Reemplazar saltos de linea internos por espacio (no se toca dobles \n\n)
    # linea simple -> espacio
     
    text = re.sub(r'(?<!\n)\s*\n\s*(?!\n)', ' ', text)

    # Colapsar multiples espacios /tab a uno solo
    text = re.sub(r'\s+', ' ', text)

    return text.strip() 
```

Ahora tenemos la situacion de que se puede encontrar secuencias de caracteres individuales que el OCR identifico, por ejemplo: 
'a y u d a', pero que juntos forman una palabra, la misma que esta en el documento original, entonces hay que decidir como agrupar estos caracteres para que formen un aproximado a la palabra del documento ariginal, o como segmentar una palabra muy larga que no pertenece al
diccionario, por ejemplo: 'requiereayuda', cuando en el documento original es:'requiere ayuda'. 

Se enfocara el problema como uno de segmentacion y se decidira las uniones o separaciones usando contexto probabilistico junto con el 
diccionario y la funcion suggest, que dada una palabra que no pertenece, devuelve un conjunto de palabras posibles. O sea, dado un string
decidir que secuencia de tokens es mas probable para esa ventana de texto.

Ya tengo una funcion que me da un score para un conunto de tokens, o un string, con el cual puedo comparar dicho score para diferentes tokens o string, la idea seria, tener un {contexto izquierdo} {cadidato} {contexto derecho}, candidato in Candidatos, 
y lo que se quiere es maximizar el valor de la funcion para quedarnos con el mas probable. Luego hay que profundizar en las maneras 
en las que se encontraran las anomalias en el texto, digamos que vamos recorriendo el texto, hasta ahora vamos encontrando palabras
validas, {contexto izquierdo}, nos encontramos la primera anomalia {anomalia}, podemos o bien seguir encontrando anomalias, o bien
encontrar palabras validas, hasta llegar a un segundo grupo de anomalias, luego de encontrar un grupo de palabras validas despues de 
la primera anomalia: {contexto izqueirdo} {primer grupo de anomalias} {contexto derecho} {segundo grupo de anomalias}, lo que hacemos es quedarnos con el contexto izquierdo, el primer grupo, y el contexto derecho, hay que ver el caso especial de cuando el 
primer grupo de anomalias es muy grande, y no es rentable tratarlo, entonces habria que obviar el contexto derecho, y trabajar solo con el izquierdo(pendiente a asignar el limite). 

Tambien hay que integrar la funcionalidad de spacy para etiquetar las palabras como VERB, DET, PUNCT, ADP, ADV, etc, combinando esto
con el corpus para calcular las prob.

```python

from spacy.tokens import Token
import numpy as np
from difflib import SequenceMatcher
from types import SimpleNamespace


# state1: busca anomalias manteniendo contexto izquierdo mientras vaya encontrando tuplas con True en [1], al encontrar go to state2
# state2: ha encontrado la primera anomalia, busca no anomalia, mientras vaya encontrando anomalias las agrega al grupo de anomalias, al encontrar go to state3
# state3: ha encontrado la primera no anomalia despues de la 1era anomalia, ahora va a buscar una nueva anomalia, mientras agrega a contexto derecho las no anomalias 
        # que encuentra, al encontrar go to state4
# state4: ha encontrado la primera anomalia del segundo grupo de anomalias, ahora usa el contexto izquierdo, el primer grupo de anomalias, y el contexto derecho, para
        # la correccion aproximada, una vez se determine la mejor, se forma el nuevo contexto izquierdo, que seria, el antiguo contexto izquierdo mas el grupo de 
        # anomalias correjidas mas el contexto derecho, go to state2

'''
input:
    arreglo de tuplas [token, Bool], donde token es el resultado de hacer spacy.load("spanish")(texto), y Bool es
    si el token pertenece al diccionario de hunspell

output:
    arreglo de tuplas ya correjidas
'''

class StringChecker():
    
    def __init__(self, model='es_core_news_sm', dicc_dir=("/usr/share/hunspell/es_ES.dic", "/usr/share/hunspell/es_ES.aff"), lm_model='anagrama_models/modelo_combinado_5.bin',
                 lm_left_ctx=5, lm_right_ctx=4):
        """
        lm_left_ctx:  numero maximo de palabras del contexto izquierdo que se pasan al LM.
                      Para un modelo trigrama (-o 3) el lookback efectivo es 2, con 3 se
                      cubre ese limite sin desperdiciar computo. Subir a 5 si se usa -o 5.
        lm_right_ctx: idem para el contexto derecho.
        """
        self.nlp = spacy.load(model)
        self.hobj = hunspell.HunSpell(dicc_dir[0], dicc_dir[1])
        self.lm_model = kenlm.LanguageModel(lm_model)
        self.lm_left_ctx = lm_left_ctx
        self.lm_right_ctx = lm_right_ctx


    def normalizar_cadena(self, anomaly_chain: list) -> list:
        """
        Fusiona rachas de 2+ tokens de un solo caracter alfabetico en un unico
        token-anomalia. Esto resuelve casos OCR como 'a y u d a' -> token 'ayuda'
        marcado como anomalia, para que la maquina de estados lo trate como bloque.
        
        Los tokens de un solo caracter no-alfabetico (puntuacion, numeros) se dejan
        tal cual para no perder informacion estructural.
        """
        result = []
        i = 0
        while i < len(anomaly_chain):
            token, valid = anomaly_chain[i]
            text = token.text.strip()

            # ¿Inicio de una racha de letras sueltas?
            if len(text) == 1 and text.isalpha():
                run = [text]
                j = i + 1
                while j < len(anomaly_chain):
                    next_tok, _ = anomaly_chain[j]
                    next_text = next_tok.text.strip()
                    if len(next_text) == 1 and next_text.isalpha():
                        run.append(next_text)
                        j += 1
                    else:
                        break

                if len(run) >= 2:
                    # Fusionar toda la racha como un unico token anomalia
                    merged_text = ''.join(run)
                    fake_token = SimpleNamespace(text=merged_text)
                    result.append((fake_token, False))
                    i = j
                else:
                    # Racha de 1: dejar el token original con su validez
                    result.append((token, valid))
                    i += 1
            else:
                result.append((token, valid))
                i += 1

        return result


    #state1
    def correjir_cadena(self, anomaly_chain: list):
        # Normalizar primero: fusionar letras sueltas consecutivas
        anomaly_chain = self.normalizar_cadena(anomaly_chain)

        self.anomaly_chain = anomaly_chain
        self.iterator: int = 0
        self.left_context = []
        self.right_context = []
        self._1rst_anom_group = []
        self._2nd_anom = None
        self.result = []

        iterator_ = self.iterator
        for i in range(iterator_, len(self.anomaly_chain)):
            if not self.anomaly_chain[i][1]: 
                self.iterator += 1
                self._1rst_anom_group.append(self.anomaly_chain[i])
                break
            else:
                self.iterator += 1
                self.left_context.append(self.anomaly_chain[i])
                self.result.append(self.anomaly_chain[i])

        if self.iterator == len(self.anomaly_chain):
            return self.result
        else:
            self.state2()

    
    def state2(self):
        iterator_ = self.iterator
        for i in range(iterator_, len(self.anomaly_chain)):
            if self.anomaly_chain[i][1]:
                self.iterator += 1
                self.right_context.append(self.anomaly_chain[i])
                break
            else:
                self.iterator += 1
                self._1rst_anom_group.append(self.anomaly_chain[i])
        
        self.state3()

    def state3(self):
        iterator_ = self.iterator
        for i in range(iterator_, len(self.anomaly_chain)):
            if not self.anomaly_chain[i][1]:
                self.iterator += 1
                self._2nd_anom = self.anomaly_chain[i]
                break
            else:
                self.iterator += 1
                self.right_context.append(self.anomaly_chain[i])
        
        self.state4()


    def state4(self):
        _1rst_anom_group_list = []
        for tok in self._1rst_anom_group:
            _1rst_anom_group_list.append(tok[0].text)

        anomaly_predict = []
        print("Comenzando proceso de correjir anomalias:")
        anomaly_predict_list = self.check_anomaly_group(self.left_context, self._1rst_anom_group, self.right_context)
        for word in anomaly_predict_list:
            anomaly_predict.append((SimpleNamespace(text=word), True))

        self.left_context.extend(anomaly_predict)
        self.left_context.extend(self.right_context)

        self.result.extend(anomaly_predict)
        self.result.extend(self.right_context)

        self._1rst_anom_group.clear()
        self.right_context.clear()

        self._1rst_anom_group.append(self._2nd_anom)

        if self.iterator == len(self.anomaly_chain):
            return self.result
        
        self.state2()
    
    def check_anomaly_group(self, left_context, anomaly_group, right_context):

        # Ventana de contexto limitada al maximo util para el orden del modelo n-grama
        left_context_str = self.get_str_for_tuple(left_context[-self.lm_left_ctx:])
        right_context_str = self.get_str_for_tuple(right_context[:self.lm_right_ctx])
        anomaly_group_str = self.get_str_for_tuple(anomaly_group)

        # Cadena limpia: solo letras espaniol, sin espacios (fusiona letras sueltas OCR)
        text = re.sub(r'[^a-záéíóúüñ]', '', anomaly_group_str.lower())
        n = len(text)

        dp = np.full(n + 1, float('-inf'), dtype=float)
        back = np.full(n + 1, -1, dtype=int)
        words_at = [''] * (n + 1)
        dp[0] = 0.0

        print("TEXT:", text)
        print("=" * 60)

        for i in range(1, n + 1):
            mejor_j = None
            mejor_pal = None

            for j in range(max(0, i - 20), i):
                palabra_raw = text[j:i]

                # --- Hunspell score ---
                if self.hobj.spell(palabra_raw):
                    spell_score = 10.0
                    palabra = palabra_raw
                    tipo = 'OK'
                else:
                    sugs = self.hobj.suggest(palabra_raw)[:5]
                    max_sug = float('-inf')
                    the_sug = None
                    for sug in sugs:
                        sim = self.similitud(palabra_raw, sug)
                        if sim > max_sug:
                            the_sug = sug
                            max_sug = sim

                    if the_sug is None or max_sug <= 0:
                        spell_score = -10.0
                        palabra = palabra_raw
                        tipo = 'BAD'
                    else:
                        spell_score = max_sug
                        palabra = the_sug
                        tipo = 'SUG'

                # --- LM score con palabras corregidas previas y contexto acotado ---
                prev_words_str = self.get_prev_words_str(j, back, words_at)
                anom_str = (prev_words_str + ' ' + palabra).strip()
                cadena_para_lm = (left_context_str + ' ' + anom_str + ' ' + right_context_str).strip()
                lm_score = self.lm_model.score(cadena_para_lm)

                # --- Penalizacion por segmentos muy cortos ---
                if len(palabra) <= 1:
                    length_penalty = -5.0
                elif len(palabra) == 2:
                    length_penalty = -1.0
                else:
                    length_penalty = 0.0

                total_score = spell_score * 0.3 + lm_score * 0.7 + length_penalty
                acumulado = dp[j] + total_score

                if acumulado > dp[i]:
                    dp[i] = acumulado
                    back[i] = j
                    words_at[i] = palabra
                    mejor_j = j
                    mejor_pal = palabra

            print(f"==> dp[{i}]={dp[i]:.2f} con '{mejor_pal}' desde j={mejor_j}")

        resultado = self.reconstruir_segmentacion(n, back, words_at)
        print("Reconstruccion final:", resultado)
        return resultado

    
    def reconstruir_segmentacion(self, n: int, back: np.ndarray, words_at: list) -> list[str]:
        """Reconstruye la lista de palabras corregidas usando words_at (no el texto raw)."""
        palabras = []
        i = n
        while i > 0:
            j = back[i]
            if j < 0:
                break
            palabras.append(words_at[i])
            i = j
        return list(reversed(palabras))


    def get_str_for_tuple(self, tuple_list: list[tuple[Token, bool]]) -> str:
        result = ''
        for t in tuple_list:
            result += t[0].text + ' '
        return result.strip()


    def get_prev_words_str(self, j: int, back: np.ndarray, words_at: list) -> str:
        """Devuelve un string con las palabras corregidas ya decididas hasta el indice j."""
        palabras = []
        i = j
        while i > 0:
            prev = back[i]
            if prev < 0:
                break
            palabras.append(words_at[i])
            i = prev
        return ' '.join(reversed(palabras))

    def similitud(self, a: str, b: str) -> float:
        return SequenceMatcher(None, a, b).ratio()
    


str_ch = StringChecker()  # usa lm_left_ctx=3, lm_right_ctx=3 por defecto
# str_ch = StringChecker(lm_left_ctx=5, lm_right_ctx=5)  # si entrenas con -o 5
```
