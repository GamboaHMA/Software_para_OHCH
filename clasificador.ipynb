En la via de clasificacion de documentos o prediccion del metadato 'tipo de documento', hay que analizar dos casos importantes: 
1. Cuando el texto viene ya con algunos metadatos
2. Cuando el texto viene en crudo, solo texto

O sea que se puede predecir el tipo de documento desde los metadatos que se le extrajo, o desde el propio texto de dicho documento.

1. Preprocesar metadatos: Se normaliza el texto de estas caracteristicas, se ponen todas a minusculas, se eliminan stopwords, se lematizan  los campos. Se codifican los campos catergoricos con one-hot o indices enteros https://www.altexsoft.com/blog/document-classification/
2. Vectorizar campos textuales: Se aplica TF-IDF sobre titulo + descriptores + notas cortas, concatenar con variables categoricas codificadas.
3. Entrenar los diferentes tipos de modelos y hacer comparaciones como evaluacion cruzada para quedarnos con la mejor, ya con las caracteristicas como indices enteros.
4. Predecir

Entrenar distintos modelos para la extraccion de los diferentes metadatos en caso de que entre solo el texto crudo. https://pmc.ncbi.nlm.nih.gov/articles/PMC3004227/. Con tecnicas como reconocer partes del documento donde suelen ir distintos campos, y extraer los valores con reglas ML o NLP. 

Reconocer el tipo de layout y en funcion del tipo de layout encontrado, escoger templates para la extraccion de cada tipo de layout.

Extraccion de metadatos con reglas y patrones(regex, cue words) como fechas, lugares listas de toponimos y patrones(En la ciudad de x), patrones que delatan ciertos tipos(Acta”, “Auto”, “Real Cédula”, “Oficio”, etc), se pueden definir en ExtractionRules.
